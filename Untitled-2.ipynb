## Prescriptive coverage recommender

**Goal:** given a situation, recommend the `pff_passCoverage` / `pff_passCoverageType` most likely to produce a good defensive outcome. We predict *effectiveness* (a win/lose outcome), **not** exact yardage — single-play yardage is too noisy to predict, but whether a play succeeds for the defense is learnable.

**Target — defensive success.** Using the standard football success-rate convention, the offense "succeeds" if it gains enough of the yards to go: ≥40% on 1st down, ≥60% on 2nd, and a conversion on 3rd/4th. **Defensive success** is the opposite (offense failed), and a defensive penalty counts as a defensive failure. On this data that split is ~50/50, so it's a balanced classification target.

Two models are built:

1. **ML classifier** (`HistGradientBoostingClassifier`, calibrated): learns `P(defensive success | situation, coverage)`. For a new situation it sweeps every coverage and ranks them by success probability. Inputs include the offense's `offenseFormation` and `personnelO` (known at the snap). On a held-out test set it beats the no-skill baseline: **AUC ≈ 0.59, accuracy ≈ 0.57 vs 0.50, and log-loss better than baseline** — modest but real signal, and good enough for ranking which coverage is more effective.
2. **Empirical model**: bins the situation (down × distance bucket × field zone) and reports the *actual* historical success rate (and avg yards) per coverage, with play counts. Prefer rows with larger `plays`.

```python
# Dependencies for the model (safe to re-run)
%pip install scikit-learn

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.calibration import CalibratedClassifierCV
from sklearn.frozen import FrozenEstimator
from sklearn.metrics import roc_auc_score, accuracy_score, log_loss

RNG = 42

def clock_to_seconds(s):
    """'MM:SS' -> seconds remaining in the quarter."""
    try:
        m, sec = str(s).split(":")
        return int(m) * 60 + int(sec)
    except Exception:
        return np.nan

# Load the filtered plays, playId as the index (identifier, not a feature)
candidates = [Path.cwd() / "plays_filtered.csv", Path(r"c:\Users\adeco\.ipython\plays_filtered.csv")]
fp = next((p for p in candidates if p.exists()), None)
if fp is None:
    raise FileNotFoundError("plays_filtered.csv not found - run the filter cell above first.")

df = pd.read_csv(fp).set_index("playId")
df["penaltyYards"] = df["penaltyYards"].fillna(0)   # NaN means no penalty
df["clockSeconds"] = df["gameClock"].map(clock_to_seconds)

# ---- Target: defensive success (offense failed to hit its success-rate threshold) ----
def offense_succeeded(row):
    ytg, gained = row["yardsToGo"], row["playResult"]
    if row["down"] == 1:   need = 0.4 * ytg
    elif row["down"] == 2: need = 0.6 * ytg
    else:                  need = ytg          # 3rd/4th down: must convert
    return gained >= need

df["def_success"] = (~df.apply(offense_succeeded, axis=1)).astype(int)
df.loc[df["penaltyYards"] > 0, "def_success"] = 0   # defensive penalty = failure

print(f"{len(df)} plays loaded; target = def_success (1 = good for defense)")
print(f"Defensive success rate: {df['def_success'].mean():.3f}")
df[["down", "yardsToGo", "yardlineNumber", "pff_passCoverage", "playResult", "def_success"]].head()
```

```python
# ---- ML classifier: predict P(defensive success) from situation + coverage ----
# offenseFormation and personnelO are known at the snap, so they are inputs the
# defense sees before choosing coverage.
SITUATION_NUM = ["quarter", "down", "yardsToGo", "yardlineNumber", "clockSeconds", "preSnapHomeScore"]
SITUATION_CAT = ["possessionTeam", "defensiveTeam", "offenseFormation", "personnelO"]
COVERAGE_CAT  = ["pff_passCoverage", "pff_passCoverageType"]
feature_cols  = SITUATION_NUM + SITUATION_CAT + COVERAGE_CAT

model_df = df.dropna(subset=["clockSeconds"]).copy()
# astype(str) first so missing offenseFormation/personnelO become an explicit category
for c in SITUATION_CAT + COVERAGE_CAT:
    model_df[c] = model_df[c].astype(str).astype("category")

X = model_df[feature_cols]
y = model_df["def_success"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RNG, stratify=y)
# further split the train set to hold out data for probability calibration
X_fit, X_cal, y_fit, y_cal = train_test_split(X_train, y_train, test_size=0.25, random_state=RNG, stratify=y_train)

base_clf = HistGradientBoostingClassifier(
    categorical_features=SITUATION_CAT + COVERAGE_CAT, random_state=RNG,
    max_iter=300, learning_rate=0.03, max_depth=4, min_samples_leaf=50, l2_regularization=1.0,
)
base_clf.fit(X_fit, y_fit)
# A calibrated copy gives trustworthy probability *values* for the metrics below.
# For RANKING coverages we use base_clf's raw scores: isotonic calibration maps many
# distinct scores to the same value, which would flatten the ranking.
cal_model = CalibratedClassifierCV(FrozenEstimator(base_clf), method="isotonic")
cal_model.fit(X_cal, y_cal)

proba = cal_model.predict_proba(X_test)[:, 1]
pred = (proba >= 0.5).astype(int)
auc = roc_auc_score(y_test, proba)
acc = accuracy_score(y_test, pred)
base_acc = max(y_test.mean(), 1 - y_test.mean())
ll = log_loss(y_test, proba)
base_ll = log_loss(y_test, np.full_like(proba, y_train.mean(), dtype=float))
print(f"Test AUC: {auc:.3f} (0.5 = no skill)")
print(f"Accuracy: {acc:.3f} | majority baseline: {base_acc:.3f}")
print(f"LogLoss:  {ll:.4f} | baseline: {base_ll:.4f}")

combos = model_df[COVERAGE_CAT].drop_duplicates().sort_values(COVERAGE_CAT).reset_index(drop=True)

def recommend_ml(possessionTeam, defensiveTeam, quarter, down, yardsToGo,
                 yardlineNumber, gameClock, preSnapHomeScore,
                 offenseFormation, personnelO, top_n=5):
    """Sweep every coverage for one situation; rank by predicted defensive-success probability.
    Returns the most effective coverages (highest P(success)) first."""
    base = {
        "quarter": quarter, "down": down, "yardsToGo": yardsToGo,
        "yardlineNumber": yardlineNumber, "clockSeconds": clock_to_seconds(gameClock),
        "preSnapHomeScore": preSnapHomeScore,
        "possessionTeam": possessionTeam, "defensiveTeam": defensiveTeam,
        "offenseFormation": str(offenseFormation), "personnelO": str(personnelO),
    }
    trial = combos.copy()
    for k, v in base.items():
        trial[k] = v
    for c in SITUATION_CAT + COVERAGE_CAT:
        trial[c] = pd.Categorical(trial[c], categories=X[c].cat.categories)
    trial["def_success_prob"] = base_clf.predict_proba(trial[feature_cols])[:, 1]
    return (trial.sort_values("def_success_prob", ascending=False)
                 [["pff_passCoverage", "pff_passCoverageType", "def_success_prob"]]
                 .head(top_n).reset_index(drop=True))

# Example situation (offense in SHOTGUN with 1 RB, 1 TE, 3 WR)
recommend_ml("DAL", "TB", quarter=4, down=3, yardsToGo=8,
             yardlineNumber=25, gameClock="02:00", preSnapHomeScore=14,
             offenseFormation="SHOTGUN", personnelO="1 RB, 1 TE, 3 WR")
```

```python
# ---- Empirical model: actual past results per coverage in comparable situations ----
def field_zone(yl):
    if yl <= 10: return "own_deep(1-10)"
    if yl <= 50: return "own_mid(11-50)"
    if yl <= 80: return "opp_mid(51-80)"
    return "opp_redzone(81-100)"

def dist_bucket(d):
    if d <= 3: return "short(1-3)"
    if d <= 7: return "mid(4-7)"
    return "long(8+)"

emp = df.copy()
emp["field_zone"]  = emp["yardlineNumber"].map(field_zone)
emp["dist_bucket"] = emp["yardsToGo"].map(dist_bucket)

def recommend_empirical(down, yardsToGo, yardlineNumber, min_plays=10, top_n=10):
    """Rank coverages by actual historical defensive-success rate in comparable past plays.
    Prefer rows with a larger `plays` count - small samples are unreliable."""
    fz, db = field_zone(yardlineNumber), dist_bucket(yardsToGo)
    sub = emp[(emp["down"] == down) & (emp["dist_bucket"] == db) & (emp["field_zone"] == fz)]
    g = (sub.groupby(["pff_passCoverage", "pff_passCoverageType"])
            .agg(def_success_rate=("def_success", "mean"),
                 avg_yards_allowed=("playResult", "mean"),
                 plays=("def_success", "size"))
            .reset_index())
    g = g[g["plays"] >= min_plays].sort_values("def_success_rate", ascending=False).reset_index(drop=True)
    print(f"Situation: down {down}, {db} to go, {fz}  |  {len(sub)} comparable past plays")
    return g.head(top_n)

# Example situation (same as above)
recommend_empirical(down=3, yardsToGo=8, yardlineNumber=25)
```
